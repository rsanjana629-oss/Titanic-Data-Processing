```python
import pandas as pd
from google.colab import files
import os

# Load the original Titanic dataset
file_path = '/content/titanic_data/train.csv'

# Check whether the original dataset exists
if not os.path.exists(file_path):
    print("Original dataset not found. Please upload train.csv first.")
else:
    df = pd.read_csv(file_path)

    # Handle missing values
    df['Age'] = df['Age'].fillna(df['Age'].median())
    df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

    # Remove duplicate rows
    df = df.drop_duplicates()

    # Encode gender
    df['Sex'] = df['Sex'].astype(str).str.strip().str.lower()
    df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

    # Encode Embarked
    df = pd.get_dummies(
        df, columns=['Embarked'], drop_first=True, dtype=int
    )

    # Verify the final dataset
    print("Dataset shape:", df.shape)
    print("Missing values:", df.isnull().sum().sum())
    print("Duplicate rows:", df.duplicated().sum())

    # Save the cleaned dataset
    output_path = '/content/titanic_cleaned.csv'
    df.to_csv(output_path, index=False)

    print("\nCleaned dataset saved successfully!")

    # Download the CSV file
    files.download(output_path)
```

```python
import pandas as pd
from google.colab import files

# 1. Load the original Titanic dataset
df = pd.read_csv('/content/train.csv')

print("Original dataset shape:", df.shape)

# 2. Handle missing values
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# 3. Check and remove duplicate rows
print("Duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()

# 4. Encode gender
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# 5. Encode Embarked
df = pd.get_dummies(
    df,
    columns=['Embarked'],
    drop_first=True,
    dtype=int
)

# 6. Verify the cleaned dataset
print("\nFinal dataset shape:", df.shape)
print("\nMissing values:")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

# 7. Save the cleaned dataset
df.to_csv('/content/titanic_cleaned.csv', index=False)

print("\nCleaned dataset saved successfully!")

# 8. Download the CSV file
files.download('/content/titanic_cleaned.csv')
```
